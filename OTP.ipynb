```python
# Simple shift cipher (like a basic one-time pad demo)

text = "ONETIMEPAD"
shift = 5
cipher = ""

for ch in text:
    if ch.isalpha():
        # Shift within A–Z range
        new_char = chr((ord(ch) - 65 + shift) % 26 + 65)
        cipher += new_char
    else:
        cipher += ch

print("Text  :", text)
print("Shift :", shift)
print("Cipher:", cipher)
```
